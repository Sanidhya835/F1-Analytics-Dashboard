```python
import pandas as pd
import numpy as np
```

```python
# Load all datasets

circuits = pd.read_csv("../data/raw/circuits.csv")
constructors = pd.read_csv("../data/raw/constructors.csv")
constructor_results = pd.read_csv("../data/raw/constructor_results.csv")
constructor_standings = pd.read_csv("../data/raw/constructor_standings.csv")
drivers = pd.read_csv("../data/raw/drivers.csv")
driver_standings = pd.read_csv("../data/raw/driver_standings.csv")
lap_times = pd.read_csv("../data/raw/lap_times.csv")
pit_stops = pd.read_csv("../data/raw/pit_stops.csv")
qualifying = pd.read_csv("../data/raw/qualifying.csv")
races = pd.read_csv("../data/raw/races.csv")
results = pd.read_csv("../data/raw/results.csv")
seasons = pd.read_csv("../data/raw/seasons.csv")
sprint_results = pd.read_csv("../data/raw/sprint_results.csv")
status = pd.read_csv("../data/raw/status.csv")
```

```python
# Replace \N with actual missing values
date_columns = [
    "date",
    "fp1_date",
    "fp2_date",
    "fp3_date",
    "quali_date",
    "sprint_date"
]

for col in date_columns:
    races[col] = races[col].replace("\\N", pd.NA)
    races[col] = pd.to_datetime(races[col])
```

# Data Cleaning Summary

The following data quality checks and transformations were performed:

### ✅ Missing Values
- Checked all 14 datasets.
- No missing values were found.

### ✅ Duplicate Records
- Checked all datasets for duplicate rows.
- Found 1,707 duplicate rows in the `lap_times` dataset.
- Verified that they were complete duplicates.
- Removed the duplicate rows.

### ✅ Data Type Validation
- Verified data types for all datasets.
- Converted the following columns to `datetime`:
  - `drivers.dob`
  - `races.date`
  - `races.fp1_date`
  - `races.fp2_date`
  - `races.fp3_date`
  - `races.quali_date`
  - `races.sprint_date`

### ✅ Special Value Handling
- Replaced `\N` values in race date columns with missing values (`NA`) before datetime conversion.

### ✅ Final Status
The datasets are cleaned, validated, and ready to be exported for SQL analysis.

```python
# Update the cleaned DataFrames in the dictionary
datasets["drivers"] = drivers
datasets["races"] = races
datasets["lap_times"] = lap_times

# Export all cleaned datasets
for name, df in datasets.items():
    df.to_csv(f"../data/cleaned/{name}.csv", index=False)

print("🎉 All cleaned datasets exported successfully!")
```
